### 1. OLTP, OLAP e Data Warehouse

#### Anotações da aula
- **OLTP:** voltado a sistemas transacionais.
- **OLAP:** relacionado a análises de dados.
- **Data Warehouse:** repositório analítico, organizado por temas, com histórico e rastreabilidade das cargas.
- Fluxo típico: **sistemas operacionais → ETL/ELT → Data Warehouse**.
- Modelagem dimensional: **tabelas fato e dimensão**.

#### Ajuste técnico do mentor
OLAP é um tipo de processamento analítico, não sinônimo de Data Warehouse. O DW é uma arquitetura de armazenamento e organização de dados que pode dar suporte a cargas OLAP.

### 2. Data Lake

#### Anotações da aula
- Repositório de dados em formato próximo ao original (dados nativos/brutos).
- Pode receber dados de várias fontes: logs, JSON, imagens e outros arquivos.
- Diferentes mecanismos (*engines*) podem processar os dados.
- Escalabilidade do armazenamento.
- Necessidade de catálogo, segurança e qualidade.

#### Ajuste técnico do mentor
Data Lakes podem armazenar dados **estruturados, semiestruturados e não estruturados**, em diversos estágios de tratamento; não apenas dados brutos.

### 3. Lakehouse e arquitetura medalhão

#### Anotações da aula
- **Lakehouse:** combina características de Data Lake e Data Warehouse, buscando reduzir silos.
- **Bronze:** dados brutos, muito próximos da origem.
- **Silver:** dados padronizados e integrados; tratamento de duplicidades, nulos e tipos.
- **Gold:** dados prontos para BI e outros destinos.
- **Views:** podem reduzir a necessidade de algumas cópias físicas.

#### Ajuste técnico do mentor
Cada camada representa um estágio de **refinamento e confiabilidade**, não uma escala obrigatória de tamanho ou volume. Nem todo dado nulo ou aparentemente duplicado deve ser removido automaticamente.

### 4. Tipos de dados e formatos

#### Anotações da aula
- Estruturados: dados em tabelas.
- Semiestruturados: JSON, CSV, Parquet (classificação anotada na aula).
- Não estruturados: grafos, imagens, vídeos (classificação anotada na aula).

#### Ajuste técnico do mentor
É importante separar **estrutura do dado** de **formato do arquivo**:
- **CSV:** normalmente transporta dados tabulares estruturados.
- **Parquet:** formato de arquivo colunar com esquema, comum em análises.
- **JSON:** exemplo frequente de estrutura semiestruturada.
- **Grafos:** possuem estrutura de nós e relações; não são automaticamente dados não estruturados.
- **Imagem e vídeo:** exemplos típicos de conteúdo não estruturado.

### 5. Logical Data Lake (aula 1.7)

#### Anotações da aula
- Dados permanecem em suas origens, com uma camada lógica que utiliza virtualização.
- Pode diminuir cópias e centralizar o acesso a várias origens.
- Catálogo descreve dados e metadados e apoia a governança.
- Ferramentas de BI podem consumir informações integradas.

#### Ponto de atenção
Virtualização não elimina necessariamente todas as movimentações ou cópias de dados; o comportamento depende da arquitetura e do desempenho exigido.

### 6. Data Ponds e Enterprise Data Hub (aula 1.8)

#### Anotações da aula
- **Data Pond:** repositório ou conjunto de dados associado a um domínio.
- **Enterprise Data Hub:** integração/centralização corporativa de dados de diferentes domínios.

#### Ajuste técnico do mentor
Um Enterprise Data Hub pode servir para compartilhar, integrar e governar dados corporativos, e não apenas guardá-los em um único lugar.

### 7. Arquiteturas analíticas e organização em camadas (aula 1.9)

#### Anotações da aula
- A organização em camadas ajuda a reduzir a ambiguidade.
- A arquitetura medalhão separa estágios do tratamento dos dados.

#### Resumo operacional
**Bronze → Silver → Gold** significa preservar a origem, refinar/validar os dados e produzir conjuntos voltados ao consumo.
